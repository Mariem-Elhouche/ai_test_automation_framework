````python
def create_bdd_prompt(user_story: str) -> str:

    system_instruction = """You are an expert QA Automation Engineer specialized in Behavior-Driven Development (BDD).
Your task is to generate comprehensive Gherkin feature files from user stories.

STRICT GHERKIN SYNTAX RULES:

1. TAGS:
   - Add MULTIPLE feature-level tags on the line before Feature:
     e.g. @categories @requiresLogin @regression @create
   - Each scenario must have EXACTLY TWO tags BEFORE the Scenario keyword:
     * First tag: @TC_ENTITYNAME_NUMBER (e.g. @TC_COMPANY_001, @TC_CATEGORY_001)
       ENTITYNAME = short uppercase domain name from the list below, NOT a description
     * Second tag: one category tag (e.g. @positive, @negative, @validation, @smoke, @security, @edge-case)

ALLOWED ENTITY NAMES (use these exact short names for @TC_ENTITYNAME_XXX):
  - LOGIN      (suite: Login)
  - COMPANY    (suite: Companies)
  - CATEGORY   (suite: Categories)
  - GROUP      (suite: Groups)
  - ENV        (suite: Environments)
  - GRAPHIC    (suite: Graphic Charter)
  - DOCUMENT   (suite: Documents)
  - DIGITAL    (suite: Digital Space)
  - ASSURE     (suite: Assures)
  - ATTACH     (suite: Attachments)
  - SERVICE    (suite: Services)
  - FAQ        (suite: FAQ)
  - FEATURE    (suite: Features)
  - DOMAIN     (suite: Domains)
  - ACTUALITE  (suite: Actualites)
  - AUTH       (suite: Authentication/Login)

Examples of correct tags: @TC_CATEGORY_001 @positive, @TC_COMPANY_002 @validation, @TC_LOGIN_003 @smoke
   - THE RULE IS: @tags always come BEFORE the Scenario keyword, never after, never between.

2. BACKGROUND:
   - Use Background ONLY if the same Given steps repeat in ALL scenarios
   - Background MUST be placed immediately after the Feature description, before any Scenario
   - If not all scenarios share the same preconditions, do NOT use Background

3. FEATURE DESCRIPTION:
   - Feature description MUST be a single short sentence describing the functionality
   - NEVER write "As a... I want to... So that..." inside the Feature block

4. USER STORY SCOPE:
   - Generate scenarios STRICTLY based on the acceptance criteria.
   - Only generate flows explicitly described in the acceptance criteria.
   - If the acceptance criteria specify that only valid credentials are handled, do NOT generate invalid credential scenarios.
   - Do NOT generate logout scenarios unless logout is explicitly mentioned.
   - Do NOT invent new business rules or flows.

5. FLOW CONSOLIDATION:
   - Combine all steps of a main positive flow into a SINGLE scenario.
   - Do NOT fragment the flow into multiple scenarios for each page or action.
   - Include optional steps (like "Rester connecté ?") only if mentioned in the acceptance criteria, integrated in the same scenario.

6. STEP ATOMICITY:
   - Each step must represent ONE user action or ONE system outcome.
   - Do NOT combine multiple actions in a single step.

7. TERMINOLOGY AND ASCII:
   - Use plain ASCII characters only — NO unicode escapes.
   - Use "page" for web navigation (login page, password page, dashboard page).
   - Test data (emails, passwords, accounts) should be referenced externally (e.g., "valid email") and not hard-coded.

8. USER BEHAVIOR VS TEST AUTOMATION:
   - Steps must describe USER actions and SYSTEM responses only.
   - NEVER describe automation scripts, Selenium actions, or internal test implementation.

9. PAGE NAVIGATION:
   - Clearly indicate when a user action leads to page redirection.
   - Use the pattern: USER ACTION → PAGE REDIRECTION
   - Example: When the user submits a valid email → Then the user is redirected to the password page

10. SCENARIO TYPES:
    - Generate ONLY functional business scenarios (user-facing behaviors)
    - Generate positive scenarios for all main success flows
    - Include boundary condition scenarios when relevant (e.g. empty fields)
    - Use Scenario Outline only if multiple input combinations are implied

11. GHERKIN QUALITY:
    - Use proper Gherkin keywords only: Feature, Background, Scenario, Scenario Outline, Given, When, Then, And, But, Examples
    - Given: initial context/state
    - When: action performed by the user
    - Then: expected outcome
    - And/But: continues the previous keyword — never start a step block with And or But
    - Use clear, concise, actionable language
    - Avoid redundant wording

12. OUTPUT:
    - Output ONLY valid Gherkin content
    - Do NOT add comments, explanations, or any text outside the Gherkin block
    - Your response must start directly with the global @tag then Feature

13. LANGUAGE RULE:
   - Gherkin keywords MUST remain in English: Feature, Background, Scenario, Scenario Outline, Given, When, Then, And, But, Examples
   - Step descriptions MUST be written in English
   - Example:
     Given the user is on the login page
     When the user enters a valid email
     Then the user is redirected to the dashboard:"""

    user_content = f"""Generate a complete Gherkin feature file for the following user story:

USER STORY:
{user_story}

IMPORTANT REMINDERS:
- Multiple global @tags on the line BEFORE Feature:
- Scenario tags on the line BEFORE the Scenario keyword — NEVER after, NEVER between Scenario and steps
- Each scenario has EXACTLY two tags: @TC_ENTITYNAME_XXX and one category tag
- Use short ENTITYNAME from the allowed list, NOT a description
- Combine all main positive steps in a single scenario
- Do not fragment multi-step flows into multiple scenarios
- Use one action per step
- Only generate flows described in the acceptance criteria
- Use "page" terminology for web navigation
- Test data should be referenced externally, not hard-coded
- Output only valid Gherkin, nothing else
- Gherkin keywords must remain in English
- Step descriptions must be written in English"""

    prompt = f"""<s>[INST] {system_instruction}

{user_content} [/INST]"""

    return prompt
````

````python
import os
import re
import torch
from datetime import datetime

# ==========================================================
# 1️⃣ Génération du fichier Gherkin
# ==========================================================

def generate_feature_file(user_story: str, max_new_tokens: int = 2048) -> str:
    """
    Génère un fichier .feature Gherkin à partir d'une user story.
    """
    prompt = create_bdd_prompt(user_story)

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=2048
    ).to(model.device)

    # ✅ Mesurer la longueur du prompt AVANT génération
    input_length = inputs["input_ids"].shape[1]

    print("🤖 Génération en cours...")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            temperature=0.7,
            top_p=0.9,
            top_k=50,
            do_sample=True,
            repetition_penalty=1.2,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id
        )

    # ✅ Décoder UNIQUEMENT les tokens générés (sans le prompt)
    generated_tokens = outputs[0][input_length:]
    generated_text = tokenizer.decode(generated_tokens, skip_special_tokens=True).strip()

    return clean_feature_content(generated_text)


# ==========================================================
# 2️⃣ Nettoyage du contenu — VERSION UNIQUE ET CORRECTE
# ==========================================================

def clean_feature_content(content: str) -> str:
    """Nettoie le contenu généré — extrait uniquement le bloc Gherkin valide."""

    if not content:
        return ""

    # Supprimer les balises markdown
    content = re.sub(r'```gherkin|```', '', content)

    # ✅ Couper tout ce qui précède Feature:
    match = re.search(r'(?m)^Feature:', content)
    if match:
        content = content[match.start():]
    else:
        return ""

    # ✅ Stopper ligne par ligne dès qu'on détecte du texte parasite
    valid_gherkin_lines = []
    for line in content.split('\n'):
        stripped = line.strip()

        if any(re.match(pattern, stripped, re.IGNORECASE) for pattern in [
            r'Generate a complete',
            r'Generate the \.feature',
            r'USER STORY',
            r'IMPORTANT REMINDERS',
            r'STRICT GHERKIN',
            r'^RULES:',
            r'Output ONLY',
            r'Do NOT',
            r'CORRECT:',
            r'WRONG:',
            r'Note:',
            r'In this feature',
            r'This feature file covers',
            r'\[INST\]',
            r'\[/INST\]',
        ]):
            break

        valid_gherkin_lines.append(line.rstrip())

    return '\n'.join(valid_gherkin_lines).strip()


# ==========================================================
# 3️⃣ Génération du nom basé sur la User Story
# ==========================================================

def generate_clean_filename_from_story(user_story: str) -> str:
    """Génère un nom de fichier court et intelligent à partir d'une user story."""

    if not user_story or not isinstance(user_story, str):
        return "generated_feature.feature"

    text = user_story.strip().lower()
    if not text:
        return "generated_feature.feature"

    if "acceptance criteria" in text:
        text = text.split("acceptance criteria")[0]

    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    patterns = [
        r"i want to (.+)",
        r"i need to (.+)",
        r"the system should (.+)",
        r"user can (.+)",
    ]

    for pattern in patterns:
        match = re.search(pattern, text)
        if match:
            action = match.group(1).strip()
            if action:
                words = action.split()[:3]
                if words:
                    filename = "_".join(words) + ".feature"
                    return re.sub(r'[^\w\-_\.]', '_', filename)

    stop_words = {"as", "a", "an", "the", "to", "i", "that", "so", "can", "in", "on", "at"}
    words = text.split()
    filtered_words = [w for w in words if w not in stop_words and len(w) > 2]

    if len(filtered_words) >= 3:
        filename = "_".join(filtered_words[:3]) + ".feature"
        return re.sub(r'[^\w\-_\.]', '_', filename)

    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    return f"feature_{timestamp}.feature"


# ==========================================================
# 4️⃣ Sauvegarde du fichier
# ==========================================================

def save_feature_file(content: str, user_story: str, filename: str = None) -> str:
    """Sauvegarde le contenu généré dans un fichier .feature."""

    if not content or not isinstance(content, str):
        raise ValueError("Le contenu du fichier ne peut pas être vide")

    output_dir = "/content/generated_features"
    os.makedirs(output_dir, exist_ok=True)

    if filename is None:
        try:
            filename = generate_clean_filename_from_story(user_story)
        except Exception as e:
            print(f"⚠️  Erreur nom fichier: {e}")
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            filename = f"feature_{timestamp}.feature"

    if not filename.endswith(".feature"):
        filename += ".feature"

    filename = re.sub(r'[<>:"/\\|?*]', '_', filename)
    filepath = os.path.join(output_dir, filename)

    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(content)

    print(f"✅ Fichier sauvegardé: {filepath}")
    return filepath


# ==========================================================
# 5️⃣ TEST
# ==========================================================

sample_story = """
As a backoffice administrator
I want to log in to the system with my credentials
So that I can access the admin dashboard securely

Acceptance Criteria:
- User can login with valid username and password
- Invalid credentials show error message
- Password field is masked
- After 3 failed attempts, account is locked for 15 minutes
"""

try:
    print("\n" + "="*60)
    print("🧪 TEST DE GÉNÉRATION")
    print("="*60)

    feature_content = generate_feature_file(sample_story)
    print("\n✅ Génération réussie")
    print("\n📄 Contenu généré:")
    print(feature_content)

    saved_path = save_feature_file(feature_content, sample_story)

    print("\n📂 Fichiers générés:")
    for f in os.listdir("/content/generated_features"):
        print(f"  - {f}")

except Exception as e:
    print(f"\n❌ Erreur: {e}")
    import traceback
    traceback.print_exc()
````

````python
import os
from gherkin.parser import Parser
from gherkin.errors import ParserError

def validate_gherkin_syntax(file_path):
    """
    Vérifie si un fichier .feature respecte la syntaxe Gherkin officielle.
    Retourne (True, None) si valide, (False, erreur) sinon.
    """
    parser = Parser()
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            gherkin_document = parser.parse(f.read())
        return True, "Syntaxe valide"
    except ParserError as e:
        return False, str(e)
    except Exception as e:
        return False, f"Erreur de lecture : {str(e)}"

# Exemple d'utilisation sur le dernier fichier généré
last_file = "/content/generated_features/login_with_valid.feature"
if os.path.exists(last_file):
    is_valid, message = validate_gherkin_syntax(last_file)
    print(f"Résultat du parsing pour {os.path.basename(last_file)} :")
    print(f"✅ {message}" if is_valid else f"❌ {message}")
else:
    print(f"Fichier non trouvé : {last_file}")
````
